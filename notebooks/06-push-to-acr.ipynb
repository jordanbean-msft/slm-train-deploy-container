# Build and Push Container Image to ACR

---

This notebook builds a Docker container image with your fine-tuned model and pushes it to Azure Container Registry (ACR).

## What This Notebook Does

1. **Builds Docker Image**: Creates a container with model, dependencies, and inference server
2. **Tags Image**: Assigns version tags for deployment tracking
3. **Authenticates to ACR**: Logs into your Azure Container Registry
4. **Pushes Image**: Uploads the container image to ACR
5. **Verifies Upload**: Confirms image is available in registry

## Why Containerize?

- **Portability**: Run anywhere (Azure Container Apps, AKS, local Docker)
- **Reproducibility**: Lock dependencies and environment
- **Scalability**: Easy to scale with orchestrators
- **Isolation**: Model runs in isolated environment
- **CI/CD**: Integrate with automated pipelines

## Container Components

The Docker image includes:
```
- Base image: Ubuntu with Python
- PyTorch and Hugging Face transformers
- Your fine-tuned model
- Inference server (FastAPI or similar)
- Health check endpoints
- Logging and monitoring
```

## Multi-Architecture Support

The build script supports:
- **linux/amd64**: Standard x86 servers
- **linux/arm64**: ARM-based instances (cost-efficient)

## Prerequisites

- Docker installed and running locally
- Azure Container Registry provisioned (via Terraform/azd)
- Fine-tuned and optimized model ready
- Sufficient disk space for image (~5-10GB)

## Expected Duration

- **First Build**: ~10-20 minutes (downloads base images)
- **Subsequent Builds**: ~3-5 minutes (uses cache)
- **Push to ACR**: ~5-10 minutes (depends on network)

## Troubleshooting

**Docker Build Failures**:
- Ensure Docker daemon is running
- Check Dockerfile syntax
- Verify base image is accessible
- Review build logs for dependency errors

**ACR Authentication Failed**:
- Run `az login` to authenticate
- Verify ACR exists: `az acr list`
- Check ACR permissions
- Ensure service principal has AcrPush role

**Push Failed**:
- Check network connectivity
- Verify ACR storage quota
- Ensure image size is within limits
- Try pushing from Azure Cloud Shell

**Large Image Size**:
- Use multi-stage builds
- Remove unnecessary dependencies
- Use smaller base images (alpine, slim)
- Exclude development tools from final image

## Useful Commands

```bash
# List images in ACR
az acr repository list --name <acr_name>

# Show image tags
az acr repository show-tags --name <acr_name> --repository <repo_name>

# Pull image locally
docker pull <acr_name>.azurecr.io/<repo_name>:<tag>

# Test container locally
docker run -p 8000:8000 <acr_name>.azurecr.io/<repo_name>:<tag>
```
