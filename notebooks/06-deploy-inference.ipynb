## Why Azure Container Apps?

| Feature | Benefit |
|---------|---------|
| **Serverless** | Pay only for actual usage, scale to zero |
| **Auto-scaling** | Handles traffic spikes automatically |
| **HTTPS** | Built-in SSL/TLS certificates |
| **Easy Updates** | Deploy new versions without downtime |
| **Cost-Effective** | Lower cost than always-on VMs |

## Troubleshooting

**Deployment Failed**:
- Check container logs in Azure Portal
- Verify ACR image exists and is pullable
- Ensure sufficient quota in subscription
- Review Container Apps environment configuration

**Container Not Starting**:
- Check health probe endpoints
- Verify environment variables
- Review container logs for errors
- Ensure model files are included in image

**High Latency**:
- Increase CPU/memory allocation
- Enable GPU acceleration if available
- Optimize model (quantization, ONNX)
- Increase min replica count to avoid cold starts

**Scale-to-Zero Issues**:
- Cold start latency (~10-30 seconds)
- Consider setting min_replicas=1 for prod
- Use warming techniques for critical paths
