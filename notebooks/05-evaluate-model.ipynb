## Evaluation Metrics

Configured in `configs/evaluation_config.yaml`:

| Metric | What It Measures | Good Score |
|--------|-----------------|-----------|
| **Perplexity** | How "surprised" model is by test data | Lower is better (<20) |
| **BLEU** | Translation/generation quality | Higher is better (>0.4) |
| **ROUGE** | Summarization quality | Higher is better (>0.5) |
| **Accuracy** | Exact match or classification accuracy | Higher is better (>0.8) |
